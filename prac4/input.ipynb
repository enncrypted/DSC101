```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt 
```

```python
def generate_graphs(nodes, edges, seed=None):
    """
    Generate random graphs with given nodes and edges, 
    save them as edge list files, and display them.
    """
    n = nodes
    for m in edges:
        # Create random graph
        G = nx.gnm_random_graph(nodes, m, seed)

        # Save graph as edge list
        filename = f"n={n}_m={m}.txt"
        nx.write_edgelist(G, filename)
        
        # Draw graph
        plt.figure(figsize=(5, 5))
        pos = nx.circular_layout(G)
        nx.draw(G, pos, with_labels=True, node_color='green',
                edge_color='blue', node_size=200, font_size=8)
        
        plt.title(f"Graph with n={n}, m={m}")
        print("Edges are :", G.edges())
        plt.show()
```

```python
generate_graphs(nodes=10, edges=np.arange(10,45,10), seed=5)
```

```python
generate_graphs(nodes=20, edges=np.arange(20,200,10), seed=5)
```
