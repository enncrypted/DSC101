```python
import networkx as nx
import numpy as np
from scipy.optimize import linprog
import matplotlib.pyplot as plt
from pulp import LpProblem, LpVariable, LpMinimize, lpSum, LpStatus, value
```

```python
def create_random_graph(n, m, seed=5):
    return nx.gnm_random_graph(n, m, seed=seed)
```

```python
def greedy_vertex_cover(G):
    cover = set()
    edges = G.edges()
    while edges:
        u, v = edges[0]  # take an arbitrary edge
        cover.add(u)
        cover.add(v)
        # Remove all edges incident to u or v
        edges = {e for e in edges if u not in e and v not in e}
    return cover
```

```python
def lp_vertex_cover(G):
    # Create LP problem instance
    prob = LpProblem("VertexCover", LpMinimize)

    # Create variables: x_v for each vertex v in G
    x = {v: LpVariable(f"x_{v}", lowBound=0, upBound=1) for v in G.nodes()}

    # Objective: minimize sum of x_v
    prob += lpSum(x[v] for v in G.nodes())

    # Constraints: for each edge, at least one endpoint must be in the cover
    for u, v in G.edges():
        prob += x[u] + x[v] >= 1

    # Solve the LP
    prob.solve()

    # Extract fractional solution
    fractional_solution = {v: x[v].varValue for v in G.nodes()}

    # Rounding: include vertices with x_v >= 0.5 in the cover
    lp_cover = {v for v, val in fractional_solution.items() if val >= 0.5}

    return lp_cover, fractional_solution
```

```python
# Displaying the result for each (n, m) pair
def run_experiment(n, m):
    # Create a graph with n nodes and m edges
    G = create_random_graph(n, m)

    # Solve the LP relaxation and rounding
    lp_cover, frac_soln = lp_vertex_cover(G)
    greedy_cover = greedy_vertex_cover(G)

    print(f"Edges: {m}")
    print(f"LP Cover size: {len(lp_cover)}")
    print(f"Greedy Cover size: {len(greedy_cover)}\n")
    if lp_cover is None:
        print("LP relaxation failed")
        return None

    node_colors = ['red' if node in lp_cover else 'blue' for node in G.nodes()]

    plt.figure(figsize=(6, 6))
    nx.draw(G, with_labels=True, node_color=node_colors, edge_color='gray', node_size=500, font_size=12)

    # nx.draw(G, with_labels=True, node_color='skyblue', edge_color='gray', node_size=500, font_size=12)
    plt.title(f'n={n}, m={m}')
    plt.show()

    # Output the results
    print(f"Graph with n={n} vertices and m={m} edges:")
    # print("Edges in the graph:", G.edges())
    print("LP Rounding Vertex Cover:", lp_cover)
    print("Size of LP Rounding Vertex Cover:", len(lp_cover))
    print("Fractional Solution:", frac_soln)
    print("-" * 80)

# Run the experiment
n = 10
m_values = [10, 20, 30, 40]

for m in m_values:
    run_experiment(n, m)
```
