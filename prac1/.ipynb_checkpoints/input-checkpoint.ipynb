```python
import networkx as nx
import matplotlib.pyplot as plt 
```

```python
G=nx.Graph()
```

```python
n=10
```

```python
G1 = nx.gnm_random_graph(n, 10)
G2 = nx.gnm_random_graph(n, 20)
G3 = nx.gnm_random_graph(n, 30)
G4 = nx.gnm_random_graph(n, 40)
```

```python
pos = nx.spring_layout(G1, seed=42)
nx.draw(G1, pos, with_labels=True, node_color='lightblue', edge_color='gray')
plt.savefig(f"graph_inp_1.png")
```

```python
pos = nx.spring_layout(G1, seed=42)
nx.draw(G2, pos, with_labels=True, node_color='lightblue', edge_color='gray')
plt.savefig(f"graph_inp_2.png")
```

```python
pos = nx.spring_layout(G1, seed=42)
nx.draw(G3, pos, with_labels=True, node_color='lightblue', edge_color='gray')
plt.savefig(f"graph_inp_3.png")
```

```python
pos = nx.spring_layout(G1, seed=42)
nx.draw(G4, pos, with_labels=True, node_color='lightblue', edge_color='gray')
plt.savefig(f"graph_inp_4.png")
```

```python
nx.write_adjlist(G1, "g1.txt")
nx.write_adjlist(G2, "g2.txt")
nx.write_adjlist(G3, "g3.txt")
nx.write_adjlist(G4, "g4.txt")
```
