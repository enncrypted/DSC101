```python
import networkx as nx
import matplotlib.pyplot as plt 
```

```python
G=nx.Graph()
```

```python
n=10
SEED = 5
```

```python
'''
used chatgpt to find a function in the networkx library gnm_ramdom_graph(n,m)
to generate a graph with the given number of nodes and edges.
here seed=SEED ensures that a particular graph will be generated at each run.
'''
G1 = nx.gnm_random_graph(n, 10, seed=SEED)
G2 = nx.gnm_random_graph(n, 20, seed=SEED)
G3 = nx.gnm_random_graph(n, 30, seed=SEED)
G4 = nx.gnm_random_graph(n, 40, seed=SEED)
```

```python
# the use of pos ensures that the visual representation of graph remains unchanged
pos = nx.spring_layout(G1, seed=42)
nx.draw(G1, pos, with_labels=True, node_color='green', edge_color='blue')
plt.savefig("graph_inp_1.png")
```

```python
G1.edges()
```

```python
pos = nx.spring_layout(G1, seed=42)
nx.draw(G2, pos, with_labels=True, node_color='green', edge_color='blue')
plt.savefig("graph_inp_2.png")
```
